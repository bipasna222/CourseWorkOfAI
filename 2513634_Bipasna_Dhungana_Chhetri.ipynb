#Task - 3 - Outlier Detection:
#Tasks:
#1. Identify outlier countries in South Asia based on their Score and GDP per Capita.

#2. Define outliers using the 1.5 × IQR rule.
#3. Create a scatter plot with GDP per Capita on the x-axis and Score on the y-axis, highlighting outliers in a different color.
#4. Discuss the characteristics of these outliers and their potential impact on regional averages.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Loading the CSV file
south_asia = pd.read_csv('/content/drive/MyDrive/2513634_Bipasna_Dhungana_Chhetri_Final_Assignment_AI/South_Asia.csv')

# Identify outliers based on Score and Log GDP per capita using 1.5 * IQR rule according to step 1

def find_outliers(column):
    Q1 = column.quantile(0.25)
    Q3 = column.quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = column[(column < lower) | (column > upper)]
    return outliers

score_outliers = find_outliers(south_asia['score'])
gdp_outliers = find_outliers(south_asia['Log GDP per capita'])

# Creating a scatter plot with GDP per Capita vs Score according to step 2
plt.figure(figsize=(10,6))
plt.scatter(south_asia['Log GDP per capita'], south_asia['score'], color='blue', label='Countries')

# Highlighting the  outliers in red as different colors.
for i in south_asia.index:
    if south_asia.loc[i, 'score'] in score_outliers.values or \
       south_asia.loc[i, 'Log GDP per capita'] in gdp_outliers.values:
        plt.scatter(south_asia.loc[i, 'Log GDP per capita'],
                    south_asia.loc[i, 'score'],
                    color='red', s=100)

plt.xlabel('Log GDP per Capita')
plt.ylabel('Score')
plt.title('South Asian Countries: Outlier Detection')
plt.legend(['Normal', 'Outlier'])
plt.show()

# Discuss characteristics of these outliers and their potemtial
print("Outlier countries based on Score or GDP per Capita:")
print(south_asia.loc[south_asia['score'].isin(score_outliers) | south_asia['Log GDP per capita'].isin(gdp_outliers),
                     ['Country name', 'score', 'Log GDP per capita']])
```

#6. Outlier Detection:

• Identify outlier countries in both regions based on Score and GDP per Capita.

• Plot these outliers and discuss their implications

```python
import pandas as pd
import matplotlib.pyplot as plt

# Loading  the CSV file
file_path = '/content/drive/MyDrive/2513634_Bipasna_Dhungana_Chhetri_Final_Assignment_AI/WHR-2024-5CS037.csv'
df = pd.read_csv(file_path)

# Defining the  South Asia and Middle East countries
south_asia_countries = ["Afghanistan", "Bangladesh", "Bhutan", "India",
                        "Maldives", "Nepal", "Pakistan", "Srilanka"]

middle_east_countries = ["Bahrain", "Iran", "Iraq", "Israel", "Jordan",
                         "Kuwait", "Lebanon", "Oman", "Palestine", "Qatar",
                         "Saudi Arabia", "Syria", "United Arab Emirates", "Yemen"]

# Uses of Filter datasets according to the question
south_asia = df[df['Country name'].isin(south_asia_countries)]
middle_east = df[df['Country name'].isin(middle_east_countries)]

# Functioning to find outliers using 1.5*IQR.

def find_outliers(column):
    Q1 = column.quantile(0.25)
    Q3 = column.quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    return column[(column < lower) | (column > upper)]

# Identifying the  outliers
sa_score_outliers = find_outliers(south_asia['score'])
sa_gdp_outliers = find_outliers(south_asia['Log GDP per capita'])

me_score_outliers = find_outliers(middle_east['score'])
me_gdp_outliers = find_outliers(middle_east['Log GDP per capita'])

# Using the Scatter plot for South Asia
plt.figure(figsize=(10,5))
plt.scatter(south_asia['Log GDP per capita'], south_asia['score'], color='blue', label='Countries')
plt.scatter(south_asia[south_asia['score'].isin(sa_score_outliers)]['Log GDP per capita'],
            sa_score_outliers, color='red', label='Outliers')
plt.xlabel('Log GDP per Capita')
plt.ylabel('Score')
plt.title('South Asian Countries: Outlier Detection')
plt.legend()
plt.show()

# Using the Scatter plot for Middle East
plt.figure(figsize=(10,5))
plt.scatter(middle_east['Log GDP per capita'], middle_east['score'], color='blue', label='Countries')
plt.scatter(middle_east[middle_east['score'].isin(me_score_outliers)]['Log GDP per capita'],
            me_score_outliers, color='red', label='Outliers')
plt.xlabel('Log GDP per Capita')
plt.ylabel('Score')
plt.title('Middle Eastern Countries: Outlier Detection')
plt.legend()
plt.show()

# Discussion Part at the end
print("Discussion:")
print("- The red points in the plots are the outliers based on Score or GDP per Capita.")
print("- Outliers can significantly affect the average happiness scores in the region.")
print("- Identifying outliers helps understand which countries are exceptionally high or low compared to the regional trend.")
```
